W5A20 Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.


```python
n = input()

def detect_number_type(s):
    try:
        s = float(s)
        if s // 1 == s:
            return int
        else:
            return float
    except:
        return 'neither'

detect_number_type(n)
```

W5A22 Viết hàm sum_halves(s) nhận chuỗi 6 chữ số và trả về tổng của hai số 3 chữ số đầu và 3 chữ số cuối.


```python
a, b, c, d, e, f = map(int, input().split())
l = [a, b, c, d, e, f]
def sum_halves(s):
    left = 0
    right = 0
    for i in range(len(s)):
        if i < 3:
            left += s[i]
        else:
            right += s[i]
        
    return left, right
    
sum_halves(l)
```

W5A28 Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3).


```python
import math
x1, y1, x2, y2, x3, y3 = map(float, input().split())
tri = [(x1, y1), (x2, y2), (x3, y3)]
def angle(tri):
    c = ((tri[1][0] - tri[0][0]) ** 2 + (tri[1][1] - tri[0][1]) ** 2) ** 0.5
    a = ((tri[2][0] - tri[1][0]) ** 2 + (tri[2][1] - tri[1][1]) ** 2) ** 0.5
    b = ((tri[2][0] - tri[0][0]) ** 2 + (tri[2][1] - tri[0][1]) ** 2) ** 0.5
    abc_angle = math.acos((a**2 + c**2 - b**2) / (2*a*c))
    return abc_angle

angle(tri)
```

W5A30 Viết hàm tính hệ số nhị thức C(n, k).


```python
k, n = map(int, input().split())

def Ckn(k, n):
    a = 1
    b = 1
    c = 1
    for i in range(1, n + 1):
        a *= i
    for i in range(1, (n-k) + 1):
        b *= i
    for i in range(1, k + 1):
        c *= i
    return int(a/(c*b))

Ckn(k, n)
```
