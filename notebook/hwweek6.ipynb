W6A10 Viết chương trình nhập vào một danh sách các số nguyên, và số nguyên
dương k. In ra list gồm các cặp số khác nhau có tổng bằng k. Mỗi cặp là
một tuple, hai số trong tuple có thể giống nhau, thứ tự xuất hiện trong danh
sách của số thứ nhất nhỏ hơn thứ tự của số thứ hai. Các tuple trong list được
sắp xếp theo thứ tự tăng dần của phần tử đầu tiên trong tuple.

```python
n = list(map(int, input().split()))
k = int(input())
result = []
for i in range(len(n)):
    if i > 0:
        for j in range(i):
            if n[j] + n[i] == k:
                result.append((n[j], n[i]))

print(sorted(result))
```

W6A21 Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ

W6A27 Cho một dãy số nguyên, Tìm và in ra vị trí các số < x (x là giá trị cho trước).
nếu không có in ra -1

```python
def find(n, x):
    a = []
    for i in range(len(n)):
        if n[i] < x:
            a.append(i)
    if len(a) == 0:
        return -1
    else:
        return a

n = list(map(int, input().split()))
x = int(input())
find(n, x)
```

W6A28 Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm
thực sự, hay vô danh (không tăng, không giảm)

```python
up = 0
down = 0
n = list(map(int, input().split()))

for i in range(len(n)):
    if i > 0:
        if n[i] > n[i - 1]:
            up += 1
        elif n[i] < n[i - 1]:
            down += 1

if up == len(n) - 1:
    print("Tang")
elif down == len(n) - 1:
    print("Giam")
else:
    print("Vo danh")
```
